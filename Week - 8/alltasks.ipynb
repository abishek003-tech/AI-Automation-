```python
import pandas as pd
import numpy as np

# Create a mock dataset for student placement analytics
np.random.seed(42)
n_samples = 200

data = {
    'Student_ID': [f'STU{i:04d}' for i in range(1, n_samples + 1)],
    'CGPA': np.round(np.random.uniform(5.0, 10.0, n_samples), 2),
    'Attendance': np.round(np.random.uniform(60, 100, n_samples), 1),
    'Aptitude_Score': np.random.randint(40, 101, n_samples),
    'Technical_Score': np.random.randint(40, 101, n_samples),
    'Communication_Score': np.random.randint(40, 101, n_samples),
    'Department': np.random.choice(['Computer Science', 'Information Technology', 'Electronics', 'Mechanical'], n_samples),
    'Internships': np.random.choice([0, 1, 2, 3], n_samples, p=[0.4, 0.4, 0.15, 0.05]),
    'Placed': np.random.choice(['Yes', 'No'], n_samples, p=[0.6, 0.4])
}

# Introduce some null values to test the data cleaning section
df_mock = pd.DataFrame(data)
df_mock.loc[df_mock.sample(frac=0.05).index, 'CGPA'] = np.nan
df_mock.loc[df_mock.sample(frac=0.05).index, 'Department'] = np.nan

# Save the mock dataset to CSV
df_mock.to_csv('student_placement_analytics.csv', index=False)
print("Created 'student_placement_analytics.csv' successfully with mock data!")
```

```python
# ============================================================
# AI AUTOMATION - 05/10/2026
# Complete Pre-ML Data Analysis Assessment
# T1 - Statistics
# T2 - Pandas Data Cleaning
# T3 - EDA
# T4 - Independent Dataset Analysis
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# STEP 1 - LOAD DATASET
# ============================================================

# Change this to your CSV file name
file_name = "student_placement_analytics.csv"

df = pd.read_csv(file_name)

print("=" * 60)
print("DATASET LOADED")
print("=" * 60)

print(df.head())


# ============================================================
# T2 - PART B: INSPECT THE DATA
# ============================================================

print("\n" + "=" * 60)
print("T2 - DATA INSPECTION")
print("=" * 60)

# Number of rows and columns
print("\nDataset Shape:")
print(df.shape)

# Column names
print("\nColumn Names:")
print(df.columns.tolist())

# Data types
print("\nData Types:")
print(df.dtypes)

# First 5 rows
print("\nFirst 5 Rows:")
print(df.head())

# Last 5 rows
print("\nLast 5 Rows:")
print(df.tail())

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Duplicate records
print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())

# Statistical summary
print("\nStatistical Summary:")
print(df.describe())


# ============================================================
# T2 - HANDLE MISSING VALUES
# ============================================================

print("\n" + "=" * 60)
print("HANDLING MISSING VALUES")
print("=" * 60)

# Separate numerical and categorical columns
numeric_columns = df.select_dtypes(include=np.number).columns
categorical_columns = df.select_dtypes(include="object").columns

# Fill numerical missing values with median
for column in numeric_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].median())

# Fill categorical missing values with mode
for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].mode()[0])

print("Missing values handled.")

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())


# ============================================================
# T2 - REMOVE DUPLICATES
# ============================================================

print("\n" + "=" * 60)
print("REMOVING DUPLICATES")
print("=" * 60)

before = len(df)

df = df.drop_duplicates()

after = len(df)

print("Rows before removing duplicates:", before)
print("Rows after removing duplicates:", after)
print("Duplicates removed:", before - after)


# ============================================================
# T2 - VALIDATE NUMERICAL VALUES
# ============================================================

print("\n" + "=" * 60)
print("DATA VALIDATION")
print("=" * 60)

# Display possible invalid values for common student columns

if "CGPA" in df.columns:
    print("\nInvalid CGPA values:")
    print(df[(df["CGPA"] < 0) | (df["CGPA"] > 10)])

if "Attendance" in df.columns:
    print("\nInvalid Attendance values:")
    print(df[(df["Attendance"] < 0) | (df["Attendance"] > 100)])

if "Aptitude_Score" in df.columns:
    print("\nInvalid Aptitude Scores:")
    print(df[
        (df["Aptitude_Score"] < 0) |
        (df["Aptitude_Score"] > 100)
    ])

if "Technical_Score" in df.columns:
    print("\nInvalid Technical Scores:")
    print(df[
        (df["Technical_Score"] < 0) |
        (df["Technical_Score"] > 100)
    ])

if "Communication_Score" in df.columns:
    print("\nInvalid Communication Scores:")
    print(df[
        (df["Communication_Score"] < 0) |
        (df["Communication_Score"] > 100)
    ])


# ============================================================
# T1 - PART A: STATISTICS
# ============================================================

print("\n" + "=" * 60)
print("T1 - STATISTICS")
print("=" * 60)

# Select numerical columns
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:

    print("\n" + "-" * 50)
    print("Column:", column)

    # Mean
    mean_value = df[column].mean()

    # Median
    median_value = df[column].median()

    # Standard deviation
    std_value = df[column].std()

    # Percentiles
    q25 = df[column].quantile(0.25)
    q50 = df[column].quantile(0.50)
    q75 = df[column].quantile(0.75)

    print("Mean:", round(mean_value, 2))
    print("Median:", round(median_value, 2))
    print("Standard Deviation:", round(std_value, 2))

    print("25th Percentile:", round(q25, 2))
    print("50th Percentile:", round(q50, 2))
    print("75th Percentile:", round(q75, 2))

    # IQR
    IQR = q75 - q25

    # Outlier limits
    lower_limit = q25 - (1.5 * IQR)
    upper_limit = q75 + (1.5 * IQR)

    # Find outliers
    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    print("Lower Limit:", round(lower_limit, 2))
    print("Upper Limit:", round(upper_limit, 2))

    print("Number of Possible Outliers:", len(outliers))

    if len(outliers) > 0:
        print("Possible Outlier Values:")
        print(outliers[column].values)

    # Simple interpretation
    print("\nInterpretation:")

    print(
        "Average value is",
        round(mean_value, 2)
    )

    print(
        "Typical middle value is",
        round(median_value, 2)
    )

    if std_value < mean_value * 0.20:
        print("The values have relatively low variation.")
    else:
        print("The values have noticeable variation.")

    if len(outliers) > 0:
        print("Possible unusual values were detected.")
    else:
        print("No possible outliers were detected.")


# ============================================================
# T2 - FILTER RECORDS
# ============================================================

print("\n" + "=" * 60)
print("T2 - FILTERING RECORDS")
print("=" * 60)

if "CGPA" in df.columns:

    high_cgpa_students = df[df["CGPA"] >= 8]

    print("\nStudents with CGPA >= 8:")
    print(high_cgpa_students.head(10))


if "Attendance" in df.columns:

    high_attendance = df[df["Attendance"] >= 75]

    print("\nStudents with Attendance >= 75:")
    print(high_attendance.head(10))


# Multiple conditions

if "CGPA" in df.columns and "Attendance" in df.columns:

    good_students = df[
        (df["CGPA"] >= 8) &
        (df["Attendance"] >= 75)
    ]

    print("\nStudents with CGPA >= 8 and Attendance >= 75:")
    print(good_students.head(10))


# ============================================================
# T2 - GROUP AND AGGREGATE
# ============================================================

print("\n" + "=" * 60)
print("T2 - GROUP AND AGGREGATE")
print("=" * 60)

if "Department" in df.columns:

    print("\nDepartment-wise Student Count:")
    print(df.groupby("Department").size())

    if "CGPA" in df.columns:

        print("\nDepartment-wise Average CGPA:")
        print(
            df.groupby("Department")["CGPA"]
            .mean()
            .round(2)
        )

    if "Attendance" in df.columns:

        print("\nDepartment-wise Average Attendance:")
        print(
            df.groupby("Department")["Attendance"]
            .mean()
            .round(2)
        )


# ============================================================
# T3 - PART C: EDA
# ============================================================

print("\n" + "=" * 60)
print("T3 - EXPLORATORY DATA ANALYSIS")
print("=" * 60)


# ------------------------------------------------------------
# 1. Numerical columns
# ------------------------------------------------------------

print("\nNumerical Variables:")
print(df.select_dtypes(include=np.number).columns.tolist())


# ------------------------------------------------------------
# 2. Categorical columns
# ------------------------------------------------------------

print("\nCategorical Variables:")
print(df.select_dtypes(include="object").columns.tolist())


# ============================================================
# VISUALIZATION 1 - HISTOGRAM
# ============================================================

if "CGPA" in df.columns:

    plt.figure(figsize=(8, 5))

    sns.histplot(
        df["CGPA"],
        bins=10,
        kde=True
    )

    plt.title("CGPA Distribution")
    plt.xlabel("CGPA")
    plt.ylabel("Number of Students")

    plt.show()


# ============================================================
# VISUALIZATION 2 - DEPARTMENT COUNT
# ============================================================

if "Department" in df.columns:

    plt.figure(figsize=(8, 5))

    sns.countplot(
        data=df,
        x="Department"
    )

    plt.title("Students by Department")
    plt.xlabel("Department")
    plt.ylabel("Number of Students")

    plt.xticks(rotation=45)

    plt.show()


# ============================================================
# VISUALIZATION 3 - CGPA VS TECHNICAL SCORE
# ============================================================

if "CGPA" in df.columns and "Technical_Score" in df.columns:

    plt.figure(figsize=(8, 5))

    sns.scatterplot(
        data=df,
        x="CGPA",
        y="Technical_Score"
    )

    plt.title("CGPA vs Technical Score")
    plt.xlabel("CGPA")
    plt.ylabel("Technical Score")

    plt.show()


# ============================================================
# VISUALIZATION 4 - ATTENDANCE VS CGPA
# ============================================================

if "Attendance" in df.columns and "CGPA" in df.columns:

    plt.figure(figsize=(8, 5))

    sns.scatterplot(
        data=df,
        x="Attendance",
        y="CGPA"
    )

    plt.title("Attendance vs CGPA")
    plt.xlabel("Attendance")
    plt.ylabel("CGPA")

    plt.show()


# ============================================================
# VISUALIZATION 5 - INTERNSHIPS VS PLACEMENT
# ============================================================

if "Internships" in df.columns and "Placed" in df.columns:

    plt.figure(figsize=(8, 5))

    sns.countplot(
        data=df,
        x="Internships",
        hue="Placed"
    )

    plt.title("Internships vs Placement")
    plt.xlabel("Number of Internships")
    plt.ylabel("Number of Students")

    plt.show()


# ============================================================
# CORRELATION
# ============================================================

print("\n" + "=" * 60)
print("CORRELATION ANALYSIS")
print("=" * 60)

correlation = df.corr(numeric_only=True)

print(correlation.round(2))


# Correlation heatmap

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Matrix")

plt.show()


# ============================================================
# T4 - DAILY CHALLENGE
# ============================================================

print("\n" + "=" * 60)
print("T4 - INDEPENDENT DATASET ANALYSIS")
print("=" * 60)


# ------------------------------------------------------------
# Q1 - What is this dataset about?
# ------------------------------------------------------------

print("\nQ1. What is this dataset about?")

print("Number of records:", len(df))
print("Number of variables:", len(df.columns))

print("Variables:")
for column in df.columns:
    print("-", column)


# ------------------------------------------------------------
# Q2 - What data quality issues exist?
# ------------------------------------------------------------

print("\nQ2. Data Quality Issues")

print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)

print("\nStatistical Summary:")
print(df.describe())


# ------------------------------------------------------------
# Q3 - Important Patterns
# ------------------------------------------------------------

print("\nQ3. Important Patterns")

print("\nCorrelation with numerical variables:")

print(
    df.corr(numeric_only=True)
    .round(2)
)


# Department pattern

if "Department" in df.columns and "CGPA" in df.columns:

    print("\nAverage CGPA by Department:")

    print(
        df.groupby("Department")["CGPA"]
        .mean()
        .round(2)
    )


# Placement pattern

if "Placed" in df.columns:

    print("\nPlacement Distribution:")

    print(df["Placed"].value_counts())


# ------------------------------------------------------------
# Q4 - Variables useful for ML
# ------------------------------------------------------------

print("\nQ4. Variables that may be useful for ML:")

for column in df.columns:

    if column not in ["Student_ID", "Placed"]:

        print("-", column)


# ------------------------------------------------------------
# Q5 - Possible Target
# ------------------------------------------------------------

print("\nQ5. Possible Target:")

if "Placed" in df.columns:

    print("Placed")

    print(
        "Reason: Placed represents the outcome that can potentially be predicted."
    )

else:

    print(
        "No obvious target column identified automatically."
    )


# ------------------------------------------------------------
# Q6 - Regression or Classification?
# ------------------------------------------------------------

print("\nQ6. Regression or Classification?")

if "Placed" in df.columns:

    unique_values = df["Placed"].nunique()

    if df["Placed"].dtype == "object":

        print("Likely Classification")

    elif unique_values <= 10:

        print("Likely Classification")

    else:

        print("Likely Regression")

else:

    print(
        "Need to inspect the target variable before deciding."
    )


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("FINAL PRE-ML ANALYSIS SUMMARY")
print("=" * 60)

print("""
1. Dataset was loaded and inspected.
2. Missing values were checked and handled.
3. Duplicate records were removed.
4. Numerical values were validated.
5. Statistics were calculated.
6. Possible outliers were identified.
7. Records were filtered.
8. Data was grouped and aggregated.
9. EDA visualizations were created.
10. Correlations were calculated.
11. Important patterns were investigated.
12. Possible ML features were identified.
13. A possible target variable was identified.
14. Regression/Classification was considered.

This completes the Pre-ML Data Analysis workflow.
""")
```
